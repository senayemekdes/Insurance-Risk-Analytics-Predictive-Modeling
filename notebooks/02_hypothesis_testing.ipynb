```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv('../data/MachineLearningRating_v3.txt', sep='|', low_memory=False)
df.columns = df.columns.str.strip()

df['LossRatio'] = df['TotalClaims'] / df['TotalPremium'].replace(0, np.nan)
df['Margin'] = df['TotalPremium'] - df['TotalClaims']
df['HasClaim'] = df['TotalClaims'] > 0

print("=== FINAL TASK 3: HYPOTHESIS TESTING ===\n")

results = []

# H1: Provinces
gauteng = df[df['Province'] == 'Gauteng']['LossRatio'].dropna()
wc = df[df['Province'] == 'Western Cape']['LossRatio'].dropna()
_, p1 = stats.ttest_ind(gauteng, wc, equal_var=False)

results.append({
    'Hypothesis': 'H1: No risk differences across provinces',
    'Groups': 'Gauteng vs Western Cape',
    'Metric': 'Loss Ratio',
    'p-value': round(p1, 4),
    'Decision': 'Reject H0' if p1 < 0.05 else 'Fail to reject H0',
    'Business Insight': 'Gauteng shows higher risk than Western Cape (p=0.0589). Although not statistically significant at 5%, it warrants further monitoring and possible regional premium adjustments.'
})

# H4: Gender
male = df[df['Gender'] == 'Male']['LossRatio'].dropna()
female = df[df['Gender'] == 'Female']['LossRatio'].dropna()
_, p4 = stats.ttest_ind(male, female, equal_var=False)

results.append({
    'Hypothesis': 'H4: No significant risk difference between Women and Men',
    'Groups': 'Male vs Female',
    'Metric': 'Loss Ratio',
    'p-value': round(p4, 4),
    'Decision': 'Reject H0' if p4 < 0.05 else 'Fail to reject H0',
    'Business Insight': 'No statistically significant difference in risk between male and female policyholders.'
})

# Convert to DataFrame
results_df = pd.DataFrame(results)
display(results_df)

# Save results
results_df.to_csv('../data/hypothesis_results.txt', index=False)
print("\n✅ Results saved to '../data/hypothesis_results.txt'")
```
